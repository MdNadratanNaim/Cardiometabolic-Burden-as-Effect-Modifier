```python
import numpy as np
import pandas as pd
```

```python
individual_record = '../bdhs-dataset/BDIR81SV/BDIR81FL.csv'
personal_record = '../bdhs-dataset/BDPR81SV/BDPR81FL.csv'

df_ind = pd.read_csv(individual_record)
df_psn = pd.read_csv(personal_record)
display(df_ind.head())
display(df_psn.head())
```

```python
df_merged = df_ind.merge(
    df_psn,
    left_on=["V001", "V002", "V003"],
    right_on=["HV001", "HV002", "HVIDX"],
    how="left",
    validate="one_to_one"
)
df_merged.head()
```

```python
df_merged.shape
```

```python
# Population
evermarried_cols = ['V020', 'S111A', 'V501']
gender_cols = ['HV104']

# Combine
population_cols = evermarried_cols + gender_cols
```

```python
# Cardio-metabolic burder
diabetes_cols = ['SB267', 'SB236', 'SB240']
hypertension_cols = ['WBP24', 'WBP25', 'WBP16', 'WBP19']
obesity_cols = ['HA40']

# Anxiety / Depression
depression_cols = ['MTH22']
anxiety_cols = ['MTH24']

# Socio-economic Status
education_cols = ['V133', 'V149', 'V106']
wealth_cols = ['V190', 'V190A', 'V191', 'V191A']
occupation_cols = ['V714', 'V731', 'V716', 'V717', 'V704', 'V705']

# Combine
exposure_cols = diabetes_cols + hypertension_cols + obesity_cols + depression_cols + anxiety_cols + education_cols + wealth_cols + occupation_cols
```

```python
# Confounders
age_cols = ['V012', 'V013']
residence_cols = ['V024', 'V025', 'V026']
marrage_cols = ['V511', 'V511A', 'V513']
religion_cols = ['V130']
children_cols = ['V201', 'V212', 'V218', 'V219']
household_cols = ['V136', 'V137', 'V138']
sleep_cols = ['MTH10']
selfhealth_cols = ['V176']
insurance_cols = ['V481']
media_cols = ['V157', 'V158', 'V159', 'V171A']
alcohol_coils = ['V485A', 'V485B']
contraceptive_cols = ['V312', 'V313']
abortion_cols = ['V228']

decision_cols = [f'V743{i}' for i in 'ABCDEF']
abuse_cols = [f'V744{i}' for i in 'ABCDE']

smoking_cols = [f'V463{i}' for i in 'ABCDEFGHIJKLXZ']
smoking_cols += ['V463AA', 'V463AB', 'V464']

food_cols = [f'V471{i}' for i in 'ABCDEFGHI'] + ['V471CS']
food_cols += [f'V472{i}' for i in 'ABCDEFGHIJKLMNOPQRSTUVW']
food_cols += [f'V472W{i}' for i in 'ABCDE']

# Combine
confounders_cols = age_cols + residence_cols + marrage_cols + religion_cols + children_cols + household_cols + sleep_cols + selfhealth_cols + insurance_cols + media_cols + alcohol_coils + contraceptive_cols + abortion_cols + decision_cols + abuse_cols + smoking_cols + food_cols
```

```python
# All Columns
cols = population_cols + exposure_cols + confounders_cols
len(cols)
```

```python
df = df_merged[cols]
df.head()
```

```python
df.replace(r'^\s*$', np.nan, regex=True, inplace=True)
df.head()
```

```python
# Features not found in BDHS dataset
df.loc[:, df.isna().sum() == df.shape[0]].columns
```

```python
# Removing "All NaN" columns and rows
df = df.loc[:, df.isna().sum() != df.shape[0]]
df = df.loc[df.isna().sum(axis=1) != df.shape[1], :]
df.head()
```

```python
df.to_csv('../bdhs-dataset/dataset.csv', index=False)
```
