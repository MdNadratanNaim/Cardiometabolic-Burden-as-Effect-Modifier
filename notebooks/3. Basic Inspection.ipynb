```python
def parse(var: str) -> None:
    """
    Details of a variable (displayed simply)
    """
    var_dict = recode.get(var, None) or clean_recode.get(var, None)
    if not var_dict:
        print("Variable not found in the recode list!")
        return None

    print(f"Name: {var_dict['name']}")
    value = var_dict.get("value", None)
    if value is None:
        print("Value: Continuous variable")
        return None
    print(f"Categories: ")
    for i, j in value.items():
        print(f"\t{i}: {j}")
```

## Imputation

Missingness in the analytic sample was minimal and confined to one covariate: religion (n=1, 0.02%). Given the negligible proportion affected, the single missing religion value was assigned to the modal category; results were unchanged in sensitivity analyses excluding this case.

```python
mode_religion = df_depression["Religion"].mode(dropna=True).iloc[0]

df_depression["Religion_imputed"] = df_depression["Religion"].isna().astype(int)
df_anxiety["Religion_imputed"] = df_anxiety["Religion"].isna().astype(int)

df_depression["Religion"] = df_depression["Religion"].fillna(mode_religion)
df_anxiety["Religion"] = df_anxiety["Religion"].fillna(mode_religion)
```
